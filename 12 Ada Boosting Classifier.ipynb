```python
import numpy as np
import pandas as pd
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import accuracy_score, classification_report
from sklearn.datasets import load_breast_cancer
```

```python
data = load_breast_cancer()
X = pd.DataFrame(data.data, columns = data.feature_names)
y = data.target
```

```python
X
```

```python
y
```

```python
X.shape
```

```python
y.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
best_estimator = DecisionTreeClassifier(max_depth=1)
```

```python
best_estimator
```

```python
adaboost = AdaBoostClassifier(estimator = best_estimator, n_estimators=50, learning_rate=1.0, random_state=42,)
```

```python
adaboost
```

```python
adb = adaboost.fit(X_train, y_train)
```

```python
adb
```

```python
adb.score(X_test, y_test)
```

```python
y_pred = adb.predict(X_test)
```

```python
y_pred
```

```python
print("Accuracy (Basic Model):", accuracy_score(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
```

```python
param_grid = {
    'n_estimators': [50, 100, 200],
    'learning_rate': [0.01, 0.1, 1, 2],
}
```

```python
ada = AdaBoostClassifier (estimator = DecisionTreeClassifier(), random_state=42)
```

```python
ada
```

```python
grid_search = GridSearchCV(estimator = ada,
                          param_grid = param_grid,
                          cv = 5,
                          scoring ='accuracy',
                          n_jobs = -1,
                          verbose = 2)
```

```python
grid_search.fit(X_train, y_train)
```

```python
print("Best Parameters:", grid_search.best_params_)
```

```python
print("Best Score on Training Set:", grid_search.best_score_)
```

```python
best_ada = grid_search.best_estimator_
```

```python
y_pred_best = best_ada.predict(X_test)
```

```python
print('Accuracy (Tuned Model):', accuracy_score(y_test, y_pred_best))
print('Classification Report:\n', classification_report(y_test, y_pred_best))
```
